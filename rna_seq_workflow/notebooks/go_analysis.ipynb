```python
import pandas as pd
from goatools.obo_parser import GODag

# load mapping
obodag = GODag("/home/nanopore/projects/rna_seq_workflow/resources/go.obo")  # download from geneontology.org

df = pd.read_csv("/home/nanopore/projects/rna_seq_workflow/results/20250603_FCOM/functional_annotations/OT_vs_OK_combined_annotations.csv")

rows = []
for _, row in df.iterrows():
    gos = str(row["GO_terms"]).replace('"', '').split(";")
    gos = [g.strip() for g in gos if g.strip()]
    for go_id in gos:
        term = obodag[go_id].name if go_id in obodag else ""
        rows.append({
            "protein_ids": row["protein_ids"],
            "symbol": row["symbol"],
            "log2FoldChange": row["log2FoldChange"],
            "baseMean": row["baseMean"],
            "padj": row["padj"],
            "GO_id": go_id,
            "GO_term": term
        })

out = pd.DataFrame(rows)
out.to_csv("genes_with_GO_ids_and_terms.csv", index=False)
```

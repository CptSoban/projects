```python
import pandas as pd
from Bio import SeqIO

df = pd.read_csv("/home/marc/projects/rna_seq_workflow/results/DEG_analysis/new_FOXG_ASM14995v2/new_FOXG_ASM14995v2_positive_DEGs.csv")

top_500 = df["protein_ids"].head(500).tolist()

print(len(top_500))

records = list(SeqIO.parse("/home/marc/data/rna_seq_foxy/genome/GCF_000149955.1/protein.faa", "fasta"))

with open("test.fa", "w") as output_pos:
    SeqIO.write((record for record in records if record.id in top_500), output_pos, "fasta")

df2 = pd.read_csv("/home/marc/projects/rna_seq_workflow/results/DEG_analysis/new_FOXG_ASM14995v2/new_FOXG_ASM14995v2_negative_DEGs.csv")

bottom_df = df2["protein_ids"].tail(500).tolist()

with open("test2.fa", "w") as output_neg:
    SeqIO.write((record for record in records if record.id in bottom_df), output_neg, "fasta")
```

```python
import requests
import re
import os

# Step 1: Fetch the latest version number
url = "https://github.com/ebi-pf-team/interproscan"
response = requests.get(url)
if response.status_code == 200:
    match = re.search(r'href="/ebi-pf-team/interproscan/releases/tag/([0-9]\.[0-9].\-[0-9]*\.[0-9])', response.text)
    if match:
        version = match.group(1)
        print(f"Latest InterProScan version: {version}")
        
        # Step 2: Construct the download URL
        download_url = f"http://ftp.ebi.ac.uk/pub/software/unix/iprscan/5/{version}/alt/interproscan-data-{version}.tar.gz"
        print(f"Download URL: {download_url}")
        
        # Step 3: Execute the curl command to download the file
        curl_command = f"curl -O {download_url}"
        os.system(curl_command)
    else:
        print("Could not find the version number.")
else:
    print("Failed to fetch the page.")
```

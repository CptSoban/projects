```python
import requests
import re
import os

# Step 1: Fetch the latest version number
url = "https://github.com/ebi-pf-team/interproscan"
response = requests.get(url)
if response.status_code == 200:
    match = re.search(r'href="/ebi-pf-team/interproscan/releases/tag/([0-9]\.[0-9].\-[0-9]*\.[0-9])', response.text)
    if match:
        version = match.group(1)
        print(f"Latest InterProScan version: {version}")
        
        # Step 2: Construct the download URL
        download_url = f"http://ftp.ebi.ac.uk/pub/software/unix/iprscan/5/{version}/alt/interproscan-data-{version}.tar.gz"
        print(f"Download URL: {download_url}")
        
        # Step 3: Execute the curl command to download the file
        curl_command = f"curl -O {download_url}"
        os.system(curl_command)
    else:
        print("Could not find the version number.")
else:
    print("Failed to fetch the page.")
```

```python
from Bio import SeqIO

with open("/home/marc/projects/rna_seq_workflow/results/DEG_analysis/20240716_FOXY/AOK_vs_OK/AOK_vs_OK.aa", "r") as input_handle, open("/home/marc/projects/rna_seq_workflow/results/DEG_analysis/20240716_FOXY/AOK_vs_OK/AOK_vs_OK_cleaned.aa", "w") as output_handle:
    # Parse the input fasta file
    sequences = SeqIO.parse(input_handle, "fasta")
    
    # Filter sequences that end with an asterisk and do not contain an asterisk elsewhere
    filtered_sequences = (seq for seq in sequences if seq.seq.endswith("*") and seq.seq.count("*") == 1)

    # Remove the trailing asterisk from these sequences
    cleaned_sequences = (seq[:len(seq.seq) - 1] + seq.seq[-1:].replace("*", "") for seq in filtered_sequences)
    
    # Parse the input fasta file again to include sequences without a trailing asterisk
    sequences = SeqIO.parse(input_handle, "fasta")
    non_asterisk_sequences = (seq for seq in sequences if not seq.seq.endswith("*"))
    
    # Combine the cleaned sequences and non-asterisk sequences
    final_sequences = list(cleaned_sequences) + (non_asterisk_sequences)
    
    # Write the final sequences to the output fasta file
    SeqIO.write(final_sequences, output_handle, "fasta")
```

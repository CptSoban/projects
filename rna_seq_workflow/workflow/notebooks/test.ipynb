```python
import pandas as pd
from Bio import SeqIO

df = pd.read_csv("/home/marc/projects/rna_seq_workflow/results/DEG_analysis/new_FOXG_ASM14995v2/new_FOXG_ASM14995v2_positive_DEGs.csv")

top_500 = df["protein_ids"].head(500).tolist()

print(len(top_500))

records = list(SeqIO.parse("/home/marc/data/rna_seq_foxy/genome/GCF_000149955.1/protein.faa", "fasta"))

with open("test.fa", "w") as output_pos:
    SeqIO.write((record for record in records if record.id in top_500), output_pos, "fasta")

df2 = pd.read_csv("/home/marc/projects/rna_seq_workflow/results/DEG_analysis/new_FOXG_ASM14995v2/new_FOXG_ASM14995v2_negative_DEGs.csv")

bottom_df = df2["protein_ids"].tail(500).tolist()

with open("test2.fa", "w") as output_neg:
    SeqIO.write((record for record in records if record.id in bottom_df), output_neg, "fasta")
```

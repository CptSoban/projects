```python
import pandas as pd
import numpy as np

main_results = pd.read_csv("/home/marc/projects/rna_seq_workflow/results/functional_annotations/new_FOXG_ASM14995v2/new_FOXG_ASM14995v2_combined_annotations.csv")
eggnog_ko = pd.read_csv("/home/marc/projects/rna_seq_workflow/results/functional_annotations/eggnog_KOs.tsv", sep="\t")

new_df = main_results[["symbol","protein_ids","log2FC_shrinked"]].set_index("symbol")


#Mask to extract highly increased and decreased DEGs into ONE table
filter_mask = (new_df["log2FC_shrinked"] >= 2) | (new_df["log2FC_shrinked"] <= -2)

#Table containing highly increased and decreased DEGs
new_df = new_df[filter_mask]

new_df["symbol"] = new_df.index

new_df = new_df.set_index("protein_ids").join(eggnog_ko.set_index("#query"))

reconstruct_df = new_df[["KEGG_ko", "log2FC_shrinked"]]

reconstruct_df = reconstruct_df.sort_values(by=["log2FC_shrinked"], ascending=False)

#reconstruct_df = reconstruct_df[reconstruct_df['KEGG_ko'] != '-']

reconstruct_df.to_csv("reconstruct.tsv", sep="\t")

color_df = new_df[new_df['KEGG_ko'] != '-'].dropna()

color_df['KEGG_ko'] = color_df['KEGG_ko'].str.split(',')

color_df = color_df.explode("KEGG_ko")

color_df = color_df[["KEGG_ko","log2FC_shrinked"]].drop_duplicates(subset=["KEGG_ko"]).set_index("KEGG_ko")

color_df = color_df.sort_values(by=["log2FC_shrinked"], ascending=False)

color_df
#color_df.to_csv("color.csv", sep="\t", header=None)
```

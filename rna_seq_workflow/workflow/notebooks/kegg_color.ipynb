```python
import pandas as pd
import numpy as np

main_results = pd.read_csv("/home/marc/projects/rna_seq_workflow/results/functional_annotations/wesdi_denovo/wesdi_denovo_combined_annotations.csv")

new_df = main_results[["symbol","log2FoldChange","KEGG_ko"]].set_index("symbol")

reconstruct_df = new_df[["KEGG_ko"]][new_df['KEGG_ko'] != '-'].dropna()

reconstruct_df["KEGG_ko"] = reconstruct_df["KEGG_ko"].str.split(',')

reconstruct_df =reconstruct_df.explode("KEGG_ko")

reconstruct_df.to_csv("wesdi_denovo_reconstruct.tsv", sep="\t")

reconstruct_df
```

```python

color_df = new_df[new_df['KEGG_ko'] != '-'].dropna()

color_df['KEGG_ko'] = color_df['KEGG_ko'].str.split(',')

color_df = color_df.explode("KEGG_ko")

# Group by column 'A' and calculate the mean of column 'B'
color_df = color_df.groupby('KEGG_ko', as_index=False).mean()

color_df = color_df.sort_values(by=["log2FoldChange"], ascending=False)

color_df = color_df.set_index("KEGG_ko")

color_df.to_csv("wesdi_denovo_color.csv", sep="\t", header=None)

color_df
```
